## Lab 1B: Inspeccionar el cluster actual

Se consulta la configuracion real del cluster al que esta conectado este notebook, no una ficha tecnica generica.

```python
cluster_id = spark.conf.get("spark.databricks.clusterUsageTags.clusterId")
cluster_info = w.clusters.get(cluster_id)

print("Cluster ID:", cluster_id)
print("Nombre:", cluster_info.cluster_name)
print("Databricks Runtime:", cluster_info.spark_version)
print("Node type:", cluster_info.node_type_id)
print("Autoscaling:", cluster_info.autoscale)
print("Num workers:", cluster_info.num_workers)
```
